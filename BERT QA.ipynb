## BERT for QA

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

tokenizer = AutoTokenizer.from_pretrained("deepset/bert-base-cased-squad2")
model = AutoModelForQuestionAnswering.from_pretrained("deepset/bert-base-cased-squad2")
context = """
Cairo is the capital city of Egypt and one of the largest cities in Africa.
It is located along the Nile River and has a rich history that dates back thousands of years.
Cairo is known for its proximity to the Giza pyramids, which are among the most famous ancient monuments in the world.
The city is a cultural and economic center, hosting many museums, universities, and businesses.
Millions of tourists visit Cairo every year to explore its historical landmarks and vibrant markets.
"""
print("BERT Question Answering System")
```

## Questions to use

```python
# Where is Cairo located?
# Why do tourists visit Cairo?
# What are the Giza pyramids?
```

## Testing

```python
print("Type 'exit' to stop\n")

while True:
    question = input("Ask a question: ")

    if question.lower() == "exit":
        break

    inputs = tokenizer(question, context, return_tensors="pt")

    with torch.no_grad():
        outputs = model(**inputs)

    start_index = torch.argmax(outputs.start_logits)
    end_index = torch.argmax(outputs.end_logits)

    answer_tokens = inputs["input_ids"][0][start_index:end_index + 1]
    answer = tokenizer.decode(answer_tokens, skip_special_tokens=True)

    print("\nAnswer:", answer)
    print("-" * 40)
```
